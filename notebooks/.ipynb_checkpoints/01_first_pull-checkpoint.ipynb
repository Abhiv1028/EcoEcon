```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Clean up the data
df['DataValue'] = pd.to_numeric(df['DataValue'], errors='coerce').fillna(0)

# Filter out summary rows and keep only sector-to-sector flows
exclude_cols = ['Total use of products', 'Total industry output', 
                'Total intermediate', 'Value added']
exclude_rows = ['Total', 'Value added', 'Compensation', 'Gross operating']

flows = df[
    ~df['ColDescr'].isin(exclude_cols) &
    ~df['RowDescr'].str.contains('|'.join(exclude_rows), na=False) &
    (df['DataValue'] > 0)
].copy()

print(f"Flows after filtering: {len(flows)}")
print(f"Unique seller sectors: {flows['RowDescr'].nunique()}")
print(f"Unique buyer sectors:  {flows['ColDescr'].nunique()}")
```

```python
# Keep only sectors that appear on both sides — true industries
both_sides = set(flows['RowDescr'].unique()) & set(flows['ColDescr'].unique())
print(f"Sectors on both sides: {len(both_sides)}")

clean_flows = flows[
    flows['RowDescr'].isin(both_sides) &
    flows['ColDescr'].isin(both_sides)
].copy()

# Normalize: each seller's outflow as proportion of their total output
# This gives us "what fraction of sector A's sales go to sector B"
# Ecological analog: what fraction of predator A's diet is prey B
total_output = clean_flows.groupby('RowDescr')['DataValue'].sum()
clean_flows['weight'] = clean_flows.apply(
    lambda row: row['DataValue'] / total_output[row['RowDescr']], axis=1
)

print(f"\nClean flows: {len(clean_flows)}")
print("\nSample normalized flows:")
print(clean_flows[['RowDescr', 'ColDescr', 'weight']].head(10).to_string())
```

```python
# Build the directed graph
G = nx.DiGraph()

# Add edges — only keep meaningful relationships (weight > 1%)
# This filters noise, same as ecologists filter weak feeding relationships
for _, row in clean_flows[clean_flows['weight'] > 0.01].iterrows():
    G.add_edge(row['RowDescr'], row['ColDescr'], weight=row['weight'])

print(f"Nodes (sectors): {G.number_of_nodes()}")
print(f"Edges (dependencies): {G.number_of_edges()}")

# ---- THE KEYSTONE ANALYSIS ----
# Betweenness centrality = how often a sector sits on the path between others
# High betweenness = keystone species = systemically important institution

betweenness = nx.betweenness_centrality(G, weight='weight')
in_degree   = nx.in_degree_centrality(G)
out_degree  = nx.out_degree_centrality(G)

# Build a summary dataframe
centrality_df = pd.DataFrame({
    'sector': list(betweenness.keys()),
    'betweenness': list(betweenness.values()),
    'in_degree':   [in_degree[s] for s in betweenness.keys()],
    'out_degree':  [out_degree[s] for s in betweenness.keys()],
}).sort_values('betweenness', ascending=False)

print("\n--- TOP 10 KEYSTONE SECTORS (2006) ---")
print(centrality_df.head(10).to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# --- LEFT: Bar chart of top 15 keystone sectors ---
ax = axes[0]
top15 = centrality_df.head(15).copy()
colors = ['#D85A30' if i < 3 else '#1D9E75' if i < 7 else '#888780' 
          for i in range(len(top15))]

bars = ax.barh(range(len(top15)), top15['betweenness'], color=colors)
ax.set_yticks(range(len(top15)))
ax.set_yticklabels([s[:45] for s in top15['sector']], fontsize=9)
ax.invert_yaxis()
ax.set_xlabel('Betweenness centrality (keystone score)')
ax.set_title('Keystone sectors — US economy 2006', fontweight='normal', fontsize=12)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Legend
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='#D85A30', label='Critical keystones'),
    Patch(facecolor='#1D9E75', label='Important hubs'),
    Patch(facecolor='#888780', label='Secondary nodes')
]
ax.legend(handles=legend_elements, fontsize=9)

# --- RIGHT: Network graph colored by betweenness ---
ax2 = axes[1]

# Layout
pos = nx.spring_layout(G, k=2, seed=42)

# Color and size nodes by betweenness
node_list = list(G.nodes())
node_betweenness = [betweenness[n] for n in node_list]
max_b = max(node_betweenness)

node_colors = [plt.cm.RdYlGn_r(b / max_b) for b in node_betweenness]
node_sizes  = [300 + 4000 * (b / max_b) for b in node_betweenness]

# Draw edges (thin and transparent)
nx.draw_networkx_edges(G, pos, ax=ax2, alpha=0.08, 
                       edge_color='gray', arrows=False)

# Draw nodes
nx.draw_networkx_nodes(G, pos, ax=ax2,
                       node_color=node_colors,
                       node_size=node_sizes,
                       alpha=0.85)

# Label only the top 8 keystones
top8 = set(centrality_df.head(8)['sector'])
labels = {n: n[:20] for n in node_list if n in top8}
nx.draw_networkx_labels(G, pos, labels, ax=ax2, font_size=7, font_weight='bold')

ax2.set_title('US economy network — node size = keystone score', 
              fontweight='normal', fontsize=12)
ax2.axis('off')

plt.suptitle('US economic ecosystem — keystone sector analysis, 2006', 
             fontsize=14, fontweight='normal', y=1.01)
plt.tight_layout()
plt.savefig('../data/keystone_sectors_2006.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved.")
```

```python
# Pull data for multiple years and track keystone scores over time
years = ['2002', '2003', '2004', '2005', '2006', '2007', '2008', '2009']

keystone_over_time = {}  # year -> full centrality dataframe
spectral_gap_over_time = {}  # year -> spectral gap value

for year in years:
    print(f"Pulling {year}...", end=' ')
    
    params['Year'] = year
    response = requests.get(url, params=params)
    data = response.json()
    rows = data['BEAAPI']['Results'][0]['Data']
    df_year = pd.DataFrame(rows)
    
    df_year['DataValue'] = pd.to_numeric(df_year['DataValue'], errors='coerce').fillna(0)
    
    flows_year = df_year[
        ~df_year['ColDescr'].isin(exclude_cols) &
        ~df_year['RowDescr'].str.contains('|'.join(exclude_rows), na=False) &
        (df_year['DataValue'] > 0)
    ].copy()
    
    both = set(flows_year['RowDescr'].unique()) & set(flows_year['ColDescr'].unique())
    flows_year = flows_year[
        flows_year['RowDescr'].isin(both) & 
        flows_year['ColDescr'].isin(both)
    ].copy()
    
    total_out = flows_year.groupby('RowDescr')['DataValue'].sum()
    flows_year['weight'] = flows_year.apply(
        lambda row: row['DataValue'] / total_out[row['RowDescr']], axis=1
    )
    
    # Build graph
    G_year = nx.DiGraph()
    for _, row in flows_year[flows_year['weight'] > 0.01].iterrows():
        G_year.add_edge(row['RowDescr'], row['ColDescr'], weight=row['weight'])
    
    # Betweenness centrality
    bet = nx.betweenness_centrality(G_year, weight='weight')
    centrality_df_year = pd.DataFrame({
        'sector': list(bet.keys()),
        'betweenness': list(bet.values()),
    }).sort_values('betweenness', ascending=False)
    keystone_over_time[year] = centrality_df_year
    
    # Spectral gap — key early warning signal
    A = nx.to_numpy_array(G_year)
    eigenvalues = np.sort(np.abs(np.linalg.eigvals(A)))[::-1]
    spectral_gap = eigenvalues[0] - eigenvalues[1]
    spectral_gap_over_time[year] = spectral_gap
    
    print(f"done. Nodes: {G_year.number_of_nodes()}, Spectral gap: {spectral_gap:.4f}")

print("\nAll years complete.")
```

```python
from scipy.stats import entropy

ebi_over_time = {}

for year in years:
    # Get betweenness centrality for this year
    df_y = keystone_over_time[year]
    
    # Shannon entropy of the betweenness distribution
    # High entropy = power distributed evenly = healthy diverse ecosystem
    # Low entropy = power concentrated in few sectors = fragile monoculture
    scores = df_y['betweenness'].values
    scores = scores[scores > 0]  # remove zeros
    
    # Normalize to get a probability distribution
    probs = scores / scores.sum()
    
    # Shannon entropy
    H = entropy(probs)
    
    # Concentration penalty — how dominated is the network by top 3 sectors?
    top3_share = df_y['betweenness'].nlargest(3).sum() / df_y['betweenness'].sum()
    
    # EBI = entropy penalized by concentration
    # Higher = more biodiverse = more resilient
    ebi = H * (1 - top3_share)
    
    ebi_over_time[year] = {
        'ebi': ebi,
        'entropy': H,
        'top3_concentration': top3_share
    }
    
    print(f"{year}: EBI={ebi:.4f}  Entropy={H:.4f}  Top3 concentration={top3_share:.4f}")
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 13))

years_int = [int(y) for y in years]
ebi_vals  = [ebi_over_time[y]['ebi'] for y in years]
conc_vals = [ebi_over_time[y]['top3_concentration'] for y in years]
spec_vals = [spectral_gap_over_time[y] for y in years]

def style_ax(ax):
    ax.axvspan(2007.5, 2009.5, alpha=0.10, color='#D85A30', label='Crisis period')
    ax.set_xticks(years_int)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

# --- Panel 1: EBI ---
ax1 = axes[0]
ax1.plot(years_int, ebi_vals, color='#1D9E75', linewidth=2.5, 
         marker='o', markersize=8, zorder=3, label='EBI')
ax1.fill_between(years_int, ebi_vals, min(ebi_vals), 
                 alpha=0.08, color='#1D9E75')

# Mark the peak
peak_year = years_int[ebi_vals.index(max(ebi_vals))]
ax1.annotate(f'Peak biodiversity\n({peak_year})',
             xy=(peak_year, max(ebi_vals)),
             xytext=(peak_year + 0.3, max(ebi_vals) - 0.05),
             arrowprops=dict(arrowstyle='->', color='#1D9E75', lw=1.5),
             fontsize=9, color='#1D9E75')

# Mark the decline
ax1.annotate('Decline begins\nbefore crisis',
             xy=(2006, ebi_over_time['2006']['ebi']),
             xytext=(2003.2, 2.86),
             arrowprops=dict(arrowstyle='->', color='#888780', lw=1.5),
             fontsize=9, color='#888780')

style_ax(ax1)
ax1.set_ylabel('Economic Biodiversity Index')
ax1.set_title('Economic Biodiversity Index (EBI) — novel metric', 
              fontsize=11, fontweight='normal')
ax1.legend(fontsize=9)

# --- Panel 2: Concentration ---
ax2 = axes[1]
ax2.plot(years_int, conc_vals, color='#D85A30', linewidth=2.5,
         marker='o', markersize=8, zorder=3, label='Top-3 sector concentration')
ax2.fill_between(years_int, conc_vals, min(conc_vals),
                 alpha=0.08, color='#D85A30')

ax2.annotate('Concentration rising\nas crisis approaches',
             xy=(2007, ebi_over_time['2007']['top3_concentration']),
             xytext=(2003.2, 0.208),
             arrowprops=dict(arrowstyle='->', color='#D85A30', lw=1.5),
             fontsize=9, color='#D85A30')

style_ax(ax2)
ax2.set_ylabel('Top-3 sector share of centrality')
ax2.set_title('Keystone concentration — ecological analog: monoculture risk',
              fontsize=11, fontweight='normal')
ax2.legend(fontsize=9)

# --- Panel 3: Spectral gap ---
ax3 = axes[2]
ax3.plot(years_int, spec_vals, color='#534AB7', linewidth=2.5,
         marker='o', markersize=8, zorder=3, label='Spectral gap')
ax3.fill_between(years_int, spec_vals, min(spec_vals),
                 alpha=0.08, color='#534AB7')

ax3.annotate('Network fragmentation\nat crisis peak (2009)',
             xy=(2009, min(spec_vals)),
             xytext=(2005.5, 0.0035),
             arrowprops=dict(arrowstyle='->', color='#534AB7', lw=1.5),
             fontsize=9, color='#534AB7')

style_ax(ax3)
ax3.set_ylabel('Spectral gap')
ax3.set_title('Spectral gap — network connectivity fragility signal',
              fontsize=11, fontweight='normal')
ax3.legend(fontsize=9)

plt.suptitle('EcoEcon — biological early warning signals for the 2008 financial crisis',
             fontsize=14, fontweight='normal', y=1.01)
plt.tight_layout()
plt.savefig('../data/ecoceon_full_analysis.png', dpi=150, bbox_inches='tight')
plt.show()
print("Saved.")
```
